**Abstraction**

Abstraction involves hiding complex implementation details and exposing only the necessary features. This reduces programming complexity and effort.

```python
from abc import ABC, abstractmethod

##Abstract base class (ABC) 
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass
### To inherit the abstract base class, a abstract method must be made 
class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)
```
