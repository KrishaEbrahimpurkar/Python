```python
class Error(Exception):
    pass

class dobException(Error): ## generic error
    pass
```

```python
year = int(input("Enter DOB"))
age = 2026-year

try:
    if age <= 30 and age >=20:
        print("Age is valid so can apply for exam")
    else:
        raise dobException
except dobException:
    print("Sorry! Your age should be greater than 20 or less than 30")
```
