**Operator Overloading**

Operator Overloading allows you to define the behaviour of operators (+,-,*,etc.) for custom objects. You achieve this by overriding specific magic methods in your class.


```python
## = default method for comparison. __eq__ used to override it,etc
## Mathematical operation for vectors (vector addition not available)
class Vector:
    def __init__(self, x,y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__ (self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, other):
        return Vector(self.x * other, self.y * other)

    def __eq__(self, other):
        return self.x == other.x, self.y == other.y

    def __repr__(self):
        return f"Vector({self.x},{self.y})"

## creat objects of the Vector class 
v1 = Vector(2,3) ## self.x, self.y
v2 = Vector(4,5) ## other.x, other.y

print(v1+v2)
print(v1-v2)
print(v1*3)
    

        
```
