**Magic Methods**

Magic methods in Python, also known as dunder methods (double undercore methods), are special methods that start and end with double underscores. These methods enable you to define the behaviour of objects for built-in operations, such as arithmetic operations, comparisons, and more.

They are predefined methods in python that can be overriden to change the behaviour of the objects. Some common magic methods include:
* init: Initializes a new instance of a class.
* str: Returns a string representation of an object.
* repr: Returns an official string representation of an object.
* len: Returns the length of an object.
* getitem: Gets an item from a container.
* setitem: Sets an item in a container.

```python
class Person: 
    pass

person = Person()
dir(person) ## magic methods
```

```python
## Bsic Magic methods
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):## overriden previuos message
        return(f"{self.name} is {self.age} years old")

person = Person("Krisha", 22)
print(person)
```
