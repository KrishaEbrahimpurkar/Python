**Pandas DataFrame and Series**

Pandas is a powerful data manipulation library in python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A series os one-dimensional array-like object, while a DataFrame is two-dimensional, size-mutable, and potentially heterogenous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
# Series (1D array like object that can hold any data type. it is similar to a column in a table.)

import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
```

```python
# Create series from dictionary
data = {"a": 1, "b":2, "c":3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ["a", "b","c"]
pd.Series(data, index = index)
```

```python
# DataFrame
#  Create a dataframe from a dictionary of list

data = {
    "Name" :["Krisha","Palak","Guman"],
    "Age":[25,30,45],
    "City": ["Banglore", "Pune", "Mumbai"]
}
df= pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
#  Create a DataFrame from a list of dictionaries
data = [
    {"Name":"Krisha","Age":22, "City": "Mumbai"},
    {"Name":"Palak","Age":30, "City": "Pune"},
    {"Name":"Guman","Age":45, "City": "Banglore"}
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
