**DECORATORS**

Decorators are a powerful and flexible feature in python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or ,ethods without modiifying thier actual code.

```python
# function copy
def welcome():
    return "Welcome to the advanced Python Course"

welcome()
```

```python
# closures functions(a function inside a function)

def main_welcome():
    msg = "welcome"
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(msg)
        print("Plese learn the concepts properly")
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
# During runtime msg given
def main_welcome(msg):
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(msg)
        print("Plese learn the concepts properly")
    return sub_welcome_method()
```

```python
main_welcome("Welcome Everyone")
```

```python
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func("Welcome evryone to this tutorial")
        print("Plese learn the concepts properly")
    return sub_welcome_method()
```

```python
def main_welcome(func,lst):
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(func(lst))
        print("Plese learn the concepts properly")
    return sub_welcome_method()
```

```python
# Decorators
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func()
        print("Plese learn the concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advanced python course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome #decorator
def course_introduction(): #goes inside as a paramter
    print("This is an advanced python course")
```

```python
#  Decorator

def my_decoarator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
# decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hi():
    print("Hi!")
```
