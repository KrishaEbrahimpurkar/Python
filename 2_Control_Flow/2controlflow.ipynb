```python
#WAP to check if year is leap or not
year = int(input("Enter year to be checked for leap or not: "))
if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print("The year is a leap year")
        else:
            print("The year is not a leap year")
    else:
        print("The year is a leap year")
else:
    print("The year is not a leap year")
    
```

```python
#Simple Calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter Operator (+, -, *, /):")

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator =="*":
    result = num1 * num2    
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero is not allowed."
else:
    result = "Invalid Operation"

print(result)
```
